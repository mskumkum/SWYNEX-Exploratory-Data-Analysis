```python
import pandas as pd

url = "https://raw.githubusercontent.com/mskumkum/Titanic-Data-Preparation/main/cleaned_titanic.csv"

df = pd.read_csv(url)

print("Dataset loaded successfully!")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.countplot(data=df,x='Survived')

plt.title("Titanic Survival Count")
plt.xlabel("Survival Status (0 = No, 1 = Yes)")
plt.ylabel("Number of Passengers")

plt.show()
```

```python
sns.countplot(data=df, x='Sex', hue='Survived')

plt.title("Survival Count by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Passengers")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(data=df,x='Age',bins=20,kde=True)

plt.title("Age Distribution of Passengers")
plt.xlabel("Age")
plt.ylabel("Count")

plt.show()
```

```python
sns.countplot(data=df,x='Pclass',hue='Survived')

plt.title("Survival by Passengers Class")
plt.xlabel("Passengers Class")
plt.ylabel("Numbers of Passengers")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(data=df,x='Fare',bins=30,kde=True)
plt.title("Fare Distribution of Passengers")
plt.xlabel("Ticket Fare")
plt.ylabel("Numbers of Passengers")

plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.heatmap(
    df.select_dtypes(include='number').corr(),
    annot=True,
    cmap='coolwarm'
)

plt.title("Correlation Heatmap")

plt.show()
```

# Exploratory Data Analysis - Titanic Dataset

## Key Findings and Insights

1. **Survival Rate:** Approximately 38.38% of passengers survived, while 61.62% did not survive.

2. **Gender Analysis:** Female passengers had a higher survival rate (74.20%) compared to male passengers (18.89%).

3. **Age Distribution:** Most passengers were young adults, mainly between 20 and 30 years old.

4. **Passenger Class:** First-class passengers had better survival outcomes than third-class passengers.

5. **Fare Analysis:** Most passengers paid lower ticket fares, while a small number paid very high fares.

6. **Correlation Analysis:** Passenger class showed a negative correlation (-0.34) with survival.

7. **Fare and Survival:** Fare showed a positive correlation (0.26) with survival.

## Conclusion

Exploratory Data Analysis helped identify important patterns in the Titanic dataset. Gender, passenger class, age and ticket fare provided useful insights into passenger survival. These findings can help in understanding the factors that may influence machine learning model predictions.
